```python
import pandas as pd
import ast

df = pd.read_csv("datasets/flipkart_com-ecommerce_sample.csv")  # your dataset name

# Convert the string list to actual Python list
df['product_category_tree'] = df['product_category_tree'].apply(
    lambda x: " ".join(ast.literal_eval(x)).replace(">>", " ").replace("'", "").strip()
)
```

```python
clothing_keywords = [
    "clothing", "t-shirt", "shirt", "top", "dress", "jeans",
    "kurta", "saree", "lehenga", "hoodie", "sweater", "footwear",
    "shoe", "sandal", "pant", "skirt", "ethnic"
]

def is_clothing(row):
    cat = str(row["product_category_tree"]).lower()
    return any(word in cat for word in clothing_keywords)
```

```python
clothes = df[df.apply(is_clothing, axis=1)]
```

```python
products = df[~df.index.isin(clothes.index)]
```

```python
def clean_text_fields(dataframe):
    dataframe["product_name"] = dataframe["product_name"].fillna("")
    dataframe["description"] = dataframe["description"].fillna("")
    dataframe["brand"] = dataframe["brand"].fillna("")
    dataframe["product_category_tree"] = dataframe["product_category_tree"].fillna("")

    dataframe["combined"] = (
        dataframe["product_name"] + " " +
        dataframe["brand"] + " " +
        dataframe["product_category_tree"] + " " +
        dataframe["description"]
    )
    return dataframe

clothes = clean_text_fields(clothes)
products = clean_text_fields(products)
```

```python
clothes.to_csv("clothes_clean.csv", index=False)
products.to_csv("products_clean.csv", index=False)

print("Done! Files created:")
print("clothes_clean.csv")
print("products_clean.csv")
```

```python
import pandas as pd

# Load your dataset
movies = pd.read_csv("datasets/movies.csv")

# Keep important columns
movies = movies[[
    "id", "title", "genres", "overview", 
    "tagline", "keywords", "credits"
]]

# Fill missing text
for col in ["genres", "overview", "tagline", "keywords", "credits"]:
    movies[col] = movies[col].fillna("")

# Clean genres (replace "-" with space)
movies["genres"] = movies["genres"].str.replace("-", " ")

# Clean keywords (replace "-" with space)
movies["keywords"] = movies["keywords"].str.replace("-", " ")

# Extract 3 main actors from credits
movies["credits"] = movies["credits"].apply(lambda x: " ".join(x.split("-")[:3]))

# Remove duplicates by ID
movies = movies.drop_duplicates(subset=["id"])

# Remove duplicates by title
movies = movies.drop_duplicates(subset=["title"])

# Create combined field for embeddings
movies["combined"] = (
    movies["title"] + " " +
    movies["genres"] + " " +
    movies["overview"] + " " +
    movies["tagline"] + " " +
    movies["keywords"] + " " +
    movies["credits"]
)

# Remove duplicates by combined text
movies = movies.drop_duplicates(subset=["combined"])

# Save cleaned dataset
movies.to_csv("movies_clean.csv", index=False)

print("✔ movies_clean.csv created with duplicates removed!")
print("Total movies:", len(movies))
```

```python
import pandas as pd

food_raw = pd.read_csv("datasets/recipes.csv")  # replace filename
food_raw.head()
```

```python
food=food_raw[['RecipeId','Name','Description','RecipeInstructions']].copy()
```

```python
food.fillna({'Name':'','Description':'','RecipeInstruction':''},inplace=True)
```

```python
import re

def clean_text(t):
    t = str(t)
    t = re.sub(r'c\(|\)', '', t)         # remove c( )
    t = t.replace('"', '')               # remove quotes
    t = t.replace("', '", " ")           # replace separators
    t = t.replace("',\"", " ")
    t = t.replace(",", " ")              # remove commas
    return t

food["Description"] = food["Description"].apply(clean_text)
food["RecipeInstructions"] = food["RecipeInstructions"].apply(clean_text)
```

```python
food["combined"] = (
    food["Name"] + " " +
    food["Description"] + " " +
    food["RecipeInstructions"]
)
```

```python
food = food.drop_duplicates(subset=["Name"])
food = food.drop_duplicates(subset=["combined"])
```

```python
food.to_csv("food_clean.csv", index=False)

print("✔ food_clean.csv is ready!")
print("Total cleaned items:", len(food))
```

```python
in_video=pd.read_csv('datasets/invideos.csv')
us_video=pd.read_csv('datasets/usvideos.csv')
column_to_keep=['video_id','title','tags','description','channel_title']
in_video=in_video[column_to_keep]
us_video=us_video[column_to_keep]
for df in[in_video,us_video]:
    for col in['tags','description','title','channel_title']:
        df[col]=df[col].fillna('').astype(str)
    df['tags']=df['tags'].str.replace("|"," ")
videos=pd.concat([in_video,us_video],ignore_index=True)
videos=videos.drop_duplicates(subset=["description"])
videos['combined']=(
    videos["title"] + " " +
    videos["tags"] + " " +
    videos["description"] + " " +
    videos["channel_title"]
)
# Remove duplicates by combined
videos = videos.drop_duplicates(subset=["combined"])

# Save cleaned file
videos.to_csv("videos_clean.csv", index=False)

print("✔ videos_clean.csv created successfully!")
print("Total video entries:", len(videos))
```
